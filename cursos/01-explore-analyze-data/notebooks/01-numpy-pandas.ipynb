## Explorando arrays de datos con NumPy

Empecemos por ver algunos datos simples.

Supongamos que un profesor universitario toma una muestra de las calificaciones de una clase para analizarla.

Ejecuta el código de la siguiente celda con el botón **▶ Run** para ver los datos.

```python
data = [50,50,47,97,49,3,53,42,26,74,82,62,37,15,70,27,36,35,48,52,63,64]
print(data)
```

Los datos se cargaron en una estructura de tipo **lista** de Python, que es un buen tipo de dato para manipulación general, pero no está optimizada para análisis numérico. Para eso usaremos el paquete **NumPy**, que incluye tipos de datos y funciones específicas para trabajar con *Núm*eros en *Py*thon.

Ejecuta la siguiente celda para cargar los datos en un **array** de NumPy.

```python
import numpy as np

grades = np.array(data)
print(grades)
```

Por si te preguntas cuáles son las diferencias entre una **lista** y un **array** de NumPy, comparemos cómo se comportan estos tipos de datos al usarlos en una expresión que los multiplica por dos.

```python
print (type(data),'x 2:', data * 2)
print('---')
print (type(grades),'x 2:', grades * 2)
```

Observa que multiplicar una lista por dos crea una nueva lista del doble de longitud, repitiendo la secuencia original de elementos. Multiplicar un array de NumPy, en cambio, realiza un cálculo elemento a elemento en el que el array se comporta como un *vector*, así que terminamos con un array del mismo tamaño donde cada elemento se multiplicó por dos.

La idea clave aquí es que los arrays de NumPy están diseñados específicamente para operaciones matemáticas sobre datos numéricos, lo que los hace más útiles para análisis de datos que una lista genérica.

Es posible que hayas notado que el tipo de clase del array de NumPy anterior es **numpy.ndarray**. El **nd** indica que es una estructura que puede tener múltiples *dimensiones* (puede tener *n* dimensiones). Nuestra instancia en particular tiene una sola dimensión de calificaciones de estudiantes.

Ejecuta la siguiente celda para ver la **forma** (shape) del array.

```python
print(grades.shape)
```

La forma confirma que este array tiene una sola dimensión, que contiene 22 elementos (hay 22 calificaciones en la lista original). Puedes acceder a los elementos individuales del array por su posición ordinal (empezando en cero). Obtengamos el primer elemento (el de la posición 0).

```python
print(grades[0])
```

Ahora que ya conoces un array de NumPy, es momento de hacer algo de análisis sobre los datos de calificaciones.

Puedes aplicar agregaciones sobre los elementos del array, así que busquemos el promedio simple de las calificaciones (es decir, el valor *medio*).

```python
print(grades.mean())
```

Entonces la calificación media es de aproximadamente 50, más o menos a la mitad del rango posible de 0 a 100.

Agreguemos un segundo conjunto de datos para los mismos estudiantes. Esta vez, registraremos el número típico de horas por semana que dedicaron a estudiar.

```python
# Definir un array con las horas de estudio
study_hours = [10.0,11.5,9.0,16.0,9.25,1.0,11.5,9.0,8.5,14.5,15.5,
               13.75,9.0,8.0,15.5,8.0,9.0,6.0,10.0,12.0,12.5,12.0]

# Crear un array 2D (un array de arrays)
student_data = np.array([study_hours, grades])

# Mostrar el array
print(student_data)
```

Ahora los datos forman un array de dos dimensiones: un array de arrays. Veamos su forma (shape).

```python
# Mostrar la forma del array 2D
print(student_data.shape)
```

El array **student_data** contiene dos elementos, cada uno de los cuales es un array con 22 elementos.

Para recorrer esta estructura, necesitas especificar la posición de cada elemento en la jerarquía. Así que para encontrar el primer valor del primer array (que contiene las horas de estudio), puedes usar el siguiente código.

```python
# Mostrar el primer elemento del primer elemento
print(student_data[0][0])
```

Ahora tienes un array multidimensional que contiene tanto el tiempo de estudio como la calificación de cada estudiante, lo que te permite comparar el tiempo de estudio con la calificación obtenida.

```python
# Obtener el valor medio de cada subarray
avg_study = student_data[0].mean()
avg_grade = student_data[1].mean()

print('Promedio de horas de estudio: {:.2f}\nPromedio de calificación: {:.2f}'.format(avg_study, avg_grade))
```

## Explorando datos tabulares con Pandas

NumPy ofrece mucha de la funcionalidad y las herramientas que necesitas para trabajar con números, como arrays de valores numéricos. Sin embargo, cuando empiezas a trabajar con tablas de datos de dos dimensiones, el paquete **Pandas** ofrece una estructura más conveniente: el **DataFrame**.

Ejecuta la siguiente celda para importar la librería Pandas y crear un DataFrame con tres columnas. La primera columna es una lista con los nombres de los estudiantes, y la segunda y tercera son los arrays de NumPy con las horas de estudio y las calificaciones.

```python
import pandas as pd

df_students = pd.DataFrame({'Name': ['Dan', 'Joann', 'Pedro', 'Rosie', 'Ethan', 'Vicky', 'Frederic', 'Jimmie', 
                                     'Rhonda', 'Giovanni', 'Francesca', 'Rajab', 'Naiyana', 'Kian', 'Jenny',
                                     'Jakeem','Helena','Ismat','Anila','Skye','Daniel','Aisha'],
                            'StudyHours':student_data[0],
                            'Grade':student_data[1]})

print(df_students)
```

Fíjate que, además de las columnas que especificaste, el DataFrame incluye un *índice* que identifica de forma única a cada fila. Podríamos haber especificado el índice explícitamente y asignado cualquier valor adecuado (por ejemplo, un correo electrónico). Sin embargo, como no especificamos ninguno, se creó uno automáticamente con un valor entero único para cada fila.

### Buscar y filtrar datos en un DataFrame

Puedes usar el método **loc** del DataFrame para obtener los datos de un valor de índice específico, así:

```python
# Obtener los datos del índice 5
print(df_students.loc[5])
```

También puedes obtener los datos de un rango de valores de índice, así:

```python
# Obtener las filas con índices del 0 al 5
print(df_students.loc[0:5])
```

Además de usar **loc** para buscar filas según el índice, puedes usar **iloc** para buscar filas según su posición ordinal en el DataFrame (sin importar el índice):

```python
# Obtener los datos de las primeras cinco filas
print(df_students.iloc[0:5])
```

Observa con atención el resultado de `iloc[0:5]` y compáralo con el de `loc[0:5]` que obtuviste antes. ¿Notas la diferencia?

El método **loc** devolvió las filas cuya *etiqueta* de índice está en la lista de valores de *0* a *5*, lo que incluye *0*, *1*, *2*, *3*, *4* y *5* (seis filas). Sin embargo, **iloc** devuelve las filas en las *posiciones* incluidas en el rango 0 a 5. Como los rangos de enteros no incluyen el límite superior, esto incluye las posiciones *0*, *1*, *2*, *3* y *4* (cinco filas).

**iloc** identifica valores en un DataFrame por *posición*, lo cual se extiende más allá de las filas hacia las columnas. Por ejemplo, puedes usarlo para encontrar los valores de las columnas en las posiciones 1 y 2 de la fila 0, así:

```python
print(df_students.iloc[0,[1,2]])
```

Volvamos al método **loc** y veamos cómo funciona con columnas. Recuerda que **loc** se usa para ubicar elementos según valores de índice, no posiciones. Al no haber una columna de índice explícita, las filas de nuestro DataFrame están indexadas con valores enteros, pero las columnas se identifican por nombre:

```python
print(df_students.loc[0,'Grade'])
```

Aquí va otro truco útil. Puedes usar **loc** para encontrar filas indexadas a partir de una expresión de filtro que haga referencia a columnas por nombre (distintas del índice), así:

```python
print(df_students.loc[df_students['Name']=='Aisha'])
```

En realidad, no necesitas usar explícitamente el método **loc** para esto. Puedes aplicar directamente una expresión de filtro sobre el DataFrame, así:

```python
print(df_students[df_students['Name']=='Aisha'])
```

Y, para más variedad, puedes lograr el mismo resultado con el método **query** del DataFrame, así:

```python
print(df_students.query('Name=="Aisha"'))
```

Los tres ejemplos anteriores dejan ver una verdad algo confusa al trabajar con Pandas: muchas veces hay varias formas de lograr el mismo resultado. Otro ejemplo de esto es la manera de referirte al nombre de una columna del DataFrame. Puedes especificar el nombre de la columna como un valor de índice con nombre (como en los ejemplos `df_students['Name']` que vimos), o puedes usar la columna como si fuera una propiedad del DataFrame, así:

```python
print(df_students[df_students.Name == 'Aisha'])
```

### Cargar un DataFrame desde un archivo

Construimos el DataFrame a partir de arrays que ya teníamos. Sin embargo, en muchos escenarios reales los datos se cargan desde archivos. Reemplacemos el DataFrame de calificaciones por el contenido de un archivo de texto.

El archivo `grades.csv` ya está disponible en `../datasets/grades.csv` (se descargó del repositorio de laboratorios de Microsoft Learn). Ejecuta la siguiente celda para cargar los datos del archivo en un DataFrame.

```python
df_students = pd.read_csv('../datasets/grades.csv',delimiter=',',header='infer')
print(df_students.head())
```

El método **read_csv** del DataFrame se usa para cargar datos desde archivos de texto. Como ves en el código, puedes especificar opciones como el delimitador de columnas y si alguna fila contiene los encabezados de columna (en este caso, el delimitador es una coma y la primera fila contiene los nombres de columna; estos son los valores por defecto, así que podríamos haber omitido esos parámetros).

### Manejo de valores faltantes

Uno de los problemas más comunes con los que debe lidiar un científico de datos son los datos incompletos o faltantes. ¿Cómo sabemos si el DataFrame tiene valores faltantes? Puedes usar el método **isnull** para identificar qué valores individuales son nulos, así:

```python
print(df_students.isnull())
```

Claro que, con un DataFrame más grande, sería ineficiente revisar todas las filas y columnas una por una, así que podemos obtener la suma de valores faltantes por columna, así:

```python
print(df_students.isnull().sum())
```

Ahora sabemos que falta un valor de **StudyHours** y dos de **Grade**.

Para verlos en contexto, podemos filtrar el DataFrame para incluir solo las filas donde alguna de las columnas (eje 1 del DataFrame) sea nula.

```python
print(df_students[df_students.isnull().any(axis=1)])
```

Cuando se obtiene el DataFrame, los valores numéricos faltantes aparecen como **NaN** (*not a number*, no es un número).

Ahora que encontramos los valores nulos, ¿qué podemos hacer con ellos?

Un enfoque común es *imputar* valores de reemplazo. Por ejemplo, si falta el número de horas de estudio, podríamos simplemente asumir que el estudiante estudió una cantidad promedio de tiempo y reemplazar el valor faltante por el promedio de horas de estudio. Para esto, podemos usar el método **fillna**, así:

```python
df_students.StudyHours = df_students.StudyHours.fillna(df_students.StudyHours.mean())
print(df_students)
```

Otra opción es asegurarte de usar solo datos que sepas que son absolutamente correctos. En ese caso, puedes eliminar filas o columnas que contengan valores nulos con el método **dropna**. Por ejemplo, eliminaremos las filas (eje 0 del DataFrame) donde alguna columna contenga valores nulos:

```python
df_students = df_students.dropna(axis=0, how='any')
print(df_students)
```

### Explorar los datos del DataFrame

Ahora que limpiamos los valores faltantes, estamos listos para explorar los datos del DataFrame. Empecemos comparando el promedio de horas de estudio y de calificaciones.

```python
# Obtener el promedio de horas de estudio usando el nombre de columna como índice
mean_study = df_students['StudyHours'].mean()

# Obtener el promedio de calificación usando el nombre de columna como propiedad (¡solo para ilustrar la diferencia!)
mean_grade = df_students.Grade.mean()

# Imprimir el promedio de horas de estudio y de calificación
print('Promedio de horas de estudio semanales: {:.2f}\nPromedio de calificación: {:.2f}'.format(mean_study, mean_grade))
```

Bien, ahora filtremos el DataFrame para encontrar solo a los estudiantes que estudiaron más que el promedio.

```python
# Obtener los estudiantes que estudiaron el promedio de horas o más
print(df_students[df_students.StudyHours > mean_study])
```

Nota que el resultado filtrado es en sí mismo un DataFrame, así que puedes trabajar con sus columnas como con cualquier otro DataFrame.

Por ejemplo, busquemos la calificación promedio de los estudiantes que estudiaron más que el promedio.

```python
# ¿Cuál fue su calificación promedio?
print(df_students[df_students.StudyHours > mean_study].Grade.mean())
```

Supongamos que la calificación para aprobar el curso es 60.

Podemos usar esa información para agregar una nueva columna al DataFrame que indique si cada estudiante aprobó o no.

Primero, crearemos una **Series** de Pandas con el indicador de aprobado/reprobado (True o False), y luego la concatenaremos como una nueva columna (eje 1) del DataFrame.

```python
passes  = pd.Series(df_students['Grade'] >= 60)
df_students = pd.concat([df_students, passes.rename("Pass")], axis=1)

print(df_students)
```

Los DataFrames están diseñados para datos tabulares, y puedes usarlos para hacer muchas de las mismas operaciones de análisis de datos que harías en una base de datos relacional, como agrupar y agregar tablas de datos.

Por ejemplo, puedes usar el método **groupby** para agrupar los datos de los estudiantes según la columna **Pass** que agregamos antes, y contar cuántos nombres hay en cada grupo. En otras palabras, puedes determinar cuántos estudiantes aprobaron y cuántos reprobaron.

```python
print(df_students.groupby(df_students.Pass).Name.count())
```

Puedes agregar múltiples campos de un grupo usando cualquier función de agregación disponible. Por ejemplo, puedes encontrar el promedio de horas de estudio y de calificación para los grupos de estudiantes que aprobaron y que reprobaron el curso.

```python
print(df_students.groupby(df_students.Pass)[['StudyHours', 'Grade']].mean())
```

Los DataFrames son increíblemente versátiles y facilitan la manipulación de datos. Muchas operaciones de DataFrame devuelven una nueva copia del DataFrame, así que si quieres modificar un DataFrame pero conservar la misma variable, debes asignar el resultado de la operación a la variable existente. Por ejemplo, el siguiente código ordena los datos de los estudiantes de forma descendente por Grade y asigna el DataFrame ordenado resultante a la variable original **df_students**.

```python
# Crear un DataFrame con los datos ordenados por Grade (descendente)
df_students = df_students.sort_values('Grade', ascending=False)

# Mostrar el DataFrame
print(df_students)
```
